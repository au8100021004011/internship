```python
import pandas as pd
```

```python
data = {
    'Student': ['Arun', 'Priya', 'Kumar', 'Divya',
                'Arun', 'Ravi', 'Meena', 'Kaviya'],

    'Age': ['21', '22', '21', '23', '21', '22', None, '22'],

    'Subject': ['Python', 'Java', 'Python', 'Java',
                'Python', 'SQL', 'Python', 'SQL'],

    'Marks': ['85', '92', '78', None, '85', '105', '88', '76'],

    'Attendance': ['90', '95', '85', '88',
                   '90', '80', '92', None]
}

df = pd.DataFrame(data)

print(df)
```

```python
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')

df['Marks'] = pd.to_numeric(df['Marks'], errors='coerce')

df['Attendance'] = pd.to_numeric(
    df['Attendance'], errors='coerce'
)
```

```python
df.dtypes
```

```python
df.loc[df['Marks'] > 100, 'Marks'] = None
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())

df['Marks'] = df['Marks'].fillna(df['Marks'].median())

df['Attendance'] = df['Attendance'].fillna(
    df['Attendance'].median()
)
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```
