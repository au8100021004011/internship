```python
plt.figure(figsize=(8,5))

plt.hist(df['math score'], bins=10, edgecolor='black')

plt.title('Distribution of Mathematics Scores')
plt.xlabel('Math Score')
plt.ylabel('Number of Students')

plt.show()
```

Histogram - Math Score

The histogram represents the distribution of mathematics scores among students. It helps identify the score ranges in which most students are concentrated and understand the overall distribution of marks.

```python
df[['math score', 'reading score', 'writing score']].hist(
    figsize=(10,6), bins=10, edgecolor='black'
)

plt.tight_layout()
plt.show()
```

Histogram – Reading and Writing Scores

The histograms show the distribution of reading and writing scores. They help compare how students performed across different score ranges.

```python
gender_avg = df.groupby('gender')['math score'].mean()

plt.figure(figsize=(7,5))

plt.bar(gender_avg.index, gender_avg.values, edgecolor='black')

plt.title('Average Math Score by Gender')
plt.xlabel('Gender')
plt.ylabel('Average Math Score')

plt.show()
```

Bar Chart – Average Math Score by Gender

The bar chart compares the average mathematics scores of male and female students. It helps identify the difference in average performance between the two groups.

```python
sns.barplot(data=df, x='test preparation course',
            y='math score')

plt.title('Math Score by Test Preparation Course')
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df['math score'], df['reading score'])

plt.title('Math Score vs Reading Score')
plt.xlabel('Math Score')
plt.ylabel('Reading Score')

plt.show()
```

Scatter Plot – Math Score vs Reading Score

The scatter plot illustrates the relationship between mathematics and reading scores. An upward pattern indicates a positive relationship between the two subjects.

```python
print("Highest Math Score:", df['math score'].max())
print("Lowest Math Score:", df['math score'].min())

print("Highest Reading Score:", df['reading score'].max())
print("Lowest Reading Score:", df['reading score'].min())

print("Highest Writing Score:", df['writing score'].max())
print("Lowest Writing Score:", df['writing score'].min())
```

```python
df[['math score', 'reading score', 'writing score']].mean()
```

```python
correlation = df[
    ['math score', 'reading score', 'writing score']
].corr()

print(correlation)
```

```python
plt.figure(figsize=(7,5))

sns.heatmap(correlation, annot=True, cmap='coolwarm')

plt.title('Correlation Between Student Scores')

plt.show()
```

 Correlation Heatmap

The correlation heatmap represents the relationships between mathematics, reading and writing scores. Higher positive correlation values indicate stronger positive linear relationships.

```python
plt.figure(figsize=(8,5))

sns.boxplot(data=df[
    ['math score', 'reading score', 'writing score']
])

plt.title('Outlier Detection in Student Scores')

plt.show()
```
